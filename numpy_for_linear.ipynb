```python
import numpy as np
np.random.seed(42) #固定随机种子，实验可复现
#输入数据，3个特征
X=2*np.random.rand(100,3).astype(np.float32) #生成100行3列的随机数据，范围在0到2之间
#真实的权重
true_W=np.array([[3],[2],[-1.5]]).astype(np.float32)
#真实标签列向量，再加上随机噪声
y=X.dot(true_W)+np.random.randn(100,1) 
```

```python
#嵌套列表[[3.], [2.], [-1.5]]：
# 每一个内层[]代表数组的一行，每行仅 1 个元素，
# 因此是3 行 1 列的二维结构；
true_W
```

```python
X.shape
X.shape[1] #特征数量，即列数
```

```python
#初始化参数
W=np.random.randn(X.shape[1],1).astype(np.float32) #随机生成3行1列的权重矩阵
W
```

```python
#超参数
learning_rate=0.001 #学习率
n_iterations=10000 #迭代次数
m=len(y)#样本数量，即行数
```

```python
print('x.shape:',X.shape)
print('y.shape:',y.shape)
print('W.shape:',W.shape)
```

```python
#纯批量梯度下降BGD：每次参数更新以全部的训练样本
#loss(w)=1/2m*∑(y(i)-X(i)·W)²，对损失函数求导
#gradients=-1/m * X.T@(y-X@W)
for iteration in range(n_iterations):
    gradients=-1/m * X.T@(y-X@W) #计算梯度
    W=W-learning_rate*gradients #更新权重
    if iteration % 50 == 0:
        loss=np.mean((y - X@W) ** 2) / 2  #计算损失
        print(f"iteration {iteration}, loss {loss:.4f}")
```

```python
print("训练权重：\n",W)
print("真实权重：\n",true_W)
```

```python
import torch
import torch.nn as nn

#1.准备数据
torch.manual_seed(42) #固定随机种子，实验可复现
X=2*torch.rand(100,3).float() #生成100行3列的随机数据，数值[0,2)
true_W=torch.tensor([[3],[2],[-1.5]]) #真实权重
#生成y时不加偏置 Y=WX+noise,没有偏置项
y=X@true_W+torch.randn(100,1).float() *0.5 #真实标签列向量，再加上随机噪声
```

```python
#2.定义线性模型
in_features=3#输入特征数量
out_features=1#输出特征数量
bias=False #不使用偏置项
#nn.Linear是PyTorch中线性层,是一个简单的线性模型，
model=nn.Linear(in_features,out_features,bias=bias) 
```

```python
#3.定义损失函数和优化器
criterion=nn.MSELoss() #均方误差损失函数
optimizer=torch.optim.SGD(model.parameters(),lr=0.1) #随机梯度下降优化器
# 模型内置了可训练的参数weight和bias（如果bias=True的话），
# 不需要手动定义权重W，也不用手动更新参数
```

```python
# 4.训练模型：小批量梯度下降（SGD）每次参数更新以一个训练样本，兼顾计算和并行
for epoch in range(500):
    #前向传播 计算预测值和损失
    y_pred=model(X)
    loss=criterion(y_pred,y)
    #反向传播和优化 标准的梯度清零、反向传播、参数更新步骤
    optimizer.zero_grad() 
    loss.backward() 
    optimizer.step() 

    if epoch % 100 == 0:
        print(f"epoch {epoch}, loss {loss.item():.4f}") #loss是一个张量，用item()获取数值
```

```python
#5.训练结果
print("训练权重：\n",model.weight.data.flatten().tolist()) #model.weight是一个张量，flatten()将其展平为一维数组.
print("真实权重：\n",true_W.flatten().tolist())
```
